```python
import pandas as pd
import numpy as np
```

```python
# From list
numbers_list = [10, 20, 30, 40, 50]
series_from_list = pd.Series(numbers_list)

print("Series from list:")
print(series_from_list)

# Check index and values
print("Index:", series_from_list.index)
print("Values:", series_from_list.values)
```

```python
# From NumPy array
numbers_array = np.array([100, 200, 300, 400])
series_from_array = pd.Series(numbers_array)

print("\nSeries from NumPy array:")
print(series_from_array)

# Check index and values
print("Index:", series_from_array.index)
print("Values:", series_from_array.values)
```
