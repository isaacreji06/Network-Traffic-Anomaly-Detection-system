```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/network_data_cleaned.csv")
df.columns = df.columns.str.strip()  # drop leading/trailing spaces in column names
print(df.head())

plt.figure(figsize=(6,4))
plt.hist(df["Packet Length Mean"], bins=30, color="skyblue", edgecolor="black")
plt.title("Distribution of Packet Length Mean")
plt.xlabel("Packet Length Mean")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df["Flow Duration"], bins=30, color="lightgreen", edgecolor="black")
plt.title("Distribution of Flow Duration")
plt.xlabel("Flow Duration (μs)")
plt.ylabel("Frequency")
plt.show()

# Clean Flow Packets/s to avoid infinities before plotting
flow_packets = df["Flow Packets/s"].replace([np.inf, -np.inf], np.nan).dropna()
plt.figure(figsize=(6,4))
plt.hist(flow_packets, bins=30, color="salmon", edgecolor="black")
plt.title("Distribution of Flow Packets per Second")
plt.xlabel("Flow Packets/s")
plt.ylabel("Frequency")
plt.show()
```
